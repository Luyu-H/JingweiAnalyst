```python
import netCDF4 as nc

def print_nc_metadata(file_path):
    try:
        dataset = nc.Dataset(file_path, 'r')
        
        for attr_name in dataset.ncattrs():
            print("Global Attributes:")
            print(f"{attr_name}: {getattr(dataset, attr_name)}")
        
        for dim_name, dim in dataset.dimensions.items():
            print(f"{dim_name}: {len(dim)} {'(unlimited)' if dim.isunlimited() else ''}")
        
        for var_name, var in dataset.variables.items():
            print(f"Variable: {var_name}")
            for attr_name in var.ncattrs():
                print(f"  {attr_name} = {getattr(var, attr_name)}")
            print("-" * 40)

        dataset.close()
        
    except Exception as e:
        print(f"Error when read the file: {e}")

file_path = "/data1/wangzhaokun-20382/Foundation_Model_results/1.1/N_1960-2024.nc"
print_nc_metadata(file_path)
```
